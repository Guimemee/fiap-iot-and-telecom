**Título:** Propagação Eletromagnética, Antenas e Orçamento de Enlace (Link Budget)

> **Enunciado Original:**
> Calcular o orçamento de enlace (Link Budget) e o alcance máximo de comunicação para um enlace de rádio LoRa na frequência de 915 MHz no espaço livre utilizando a Equação de Friis. O transmissor possui potência Pt = +14 dBm e ganho de antena Gt = +2 dBi; o receptor possui ganho Gr = +3 dBi e sensibilidade de recepção Pr_min = -137 dBm. Considerar margem de desvanecimento (Fade Margin) de 15 dB e perdas em conectores/cabos de 2 dB.

### Formulação Matemática Formal
$$P_{rx} = P_{tx} + G_{tx} + G_{rx} - \text{FSPL} - L_{\text{outros}}, \quad \text{FSPL}(\text{dB}) = 20\log_{10}(d) + 20\log_{10}(f) + 32,44$$


### Passo 1: Cálculo da Atenuação Máxima Permitida (Path Loss Máximo)
Equação de balanço: \text{FSPL}_{max} = P_{tx} + G_{tx} + G_{rx} - P_{rx\_min} - L_{cabos} - \text{Margem}. Substituindo: \text{FSPL}_{max} = 14 + 2 + 3 - (-137) - 2 - 15 = 139\text{ dB}.


### Passo 2: Fórmula da Perda em Espaço Livre (FSPL)
Para frequência f em MHz (915 MHz) e distância d em quilômetros: \text{FSPL} = 20\log_{10}(d) + 20\log_{10}(915) + 32,44 = 20\log_{10}(d) + 59,23 + 32,44 = 20\log_{10}(d) + 91,67\text{ dB}.


### Passo 3: Igualação e Isolamento da Distância d
Temos 20\log_{10}(d) + 91,67 = 139 \implies 20\log_{10}(d) = 139 - 91,67 = 47,33\text{ dB}.


### Passo 4: Resolução Logarítmica do Raio de Alcance
\log_{10}(d) = \frac{47,33}{20} \approx 2,3665 \implies d = 10^{2,3665} \approx 232,5\text{ km} em condições ideais de linha de visada (LOS puro em espaço livre).


### Passo 5: Correção para Ambiente Terrestre (Modelo Hata / Okumura)
Em relevo urbano e industrial com obstruções e desvanecimento por múltiplos percursos, o expoente de perda sobe de n=2 para n=3,5 a 4, resultando em alcance prático de 12 a 18 km.


```python
import numpy as np
import matplotlib.pyplot as plt

dist_km = np.linspace(0.1, 25, 200)
f_mhz = 915
fspl_los = 20*np.log10(dist_km) + 20*np.log10(f_mhz) + 32.44
# Potencia recebida com Pt=14, Gt=2, Gr=3, L=2
prx = 14 + 2 + 3 - fspl_los - 2
sens = -137
margem = 15
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#181436')
ax.set_facecolor('#100d22')
ax.plot(dist_km, prx, color='#a78bfa', lw=2.5, label='Potência Recebida Prx (dBm)')
ax.axhline(sens, color='#ef4444', ls='--', lw=1.5, label='Sensibilidade Limite (-137 dBm)')
ax.axhline(sens + margem, color='#f59e0b', ls=':', lw=1.5, label='Limiar com Fade Margin (-122 dBm)')
ax.set_title('Orçamento de Enlace e Atenuação no Espaço Livre (915 MHz)', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Distância do Enlace (Quilômetros)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Nível de Sinal Recebido (dBm)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#3a3275', alpha=0.5, ls=':')
ax.legend(facecolor='#241f4d', edgecolor='#3a3275', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
