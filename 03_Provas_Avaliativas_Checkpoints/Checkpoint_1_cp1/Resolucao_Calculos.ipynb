### Formulação Matemática Formal
$$R_b = \text{SF} \cdot \frac{\text{BW}}{2^{\text{SF}}} \cdot \text{CR}, \quad \text{Sensibilidade} = -174 + 10\log_{10}(\text{BW}) + \text{NF} + \text{SNR}$$


### Passo 1: Espectro e Regulação Anatel
No Brasil, o LoRaWAN opera na banda ISM não licenciada de 915-928 MHz (AU915), o Sigfox em 920 MHz e o NB-IoT em bandas licenciadas de operadoras celulares (Band 28 em 700 MHz).


### Passo 2: Modulação Chirp Spread Spectrum (CSS) no LoRa
O fator de espalhamento (Spreading Factor - SF) varia de SF7 a SF12. Com maior SF, a taxa de dados diminui, mas a sensibilidade do receptor melhora drasticamente até -137 dBm.


### Passo 3: Cálculo da Taxa de Dados para SF7 e SF12
Para BW = 125 kHz e Coding Rate CR = 4/5: Para SF7: R_b = 7 \times \frac{125000}{128} \times 0,8 \approx 5,47\text{ kbps}. Para SF12: R_b = 12 \times \frac{125000}{4096} \times 0,8 \approx 0,293\text{ kbps} (293 bps).


### Passo 4: Capacidade de Carga Útil e Limites de Duty Cycle
Sigfox limitado a 12 bytes e 140 mensagens/dia; LoRaWAN suporta payloads de até 242 bytes (SF7); NB-IoT suporta pacotes de até 1600 bytes com taxa de até 200 kbps.


### Passo 5: Conclusão de Aplicação Industrial
Para sensores fabris e de agronegócio distribuídos sem infraestrutura celular, o LoRaWAN com gateway privado oferece o menor custo total de propriedade (TCO).


```python
import numpy as np
import matplotlib.pyplot as plt

sf = np.arange(7, 13)
bw = 125000
cr = 4 / 5
rb_bps = sf * (bw / (2**sf)) * cr
sens_dbm = -174 + 10*np.log10(bw) + 6 + (-7.5 - 2.5*(sf - 7))
fig, ax1 = plt.subplots(figsize=(6, 3.5), facecolor='#181436')
ax1.set_facecolor('#100d22')
ax2 = ax1.twinx()
p1 = ax1.plot(sf, rb_bps / 1000, marker='o', color='#8b5cf6', lw=2.5, label='Taxa de Dados (kbps)')
p2 = ax2.plot(sf, sens_dbm, marker='s', color='#22d3ee', lw=2.5, label='Sensibilidade RF (dBm)')
ax1.set_xlabel('Spreading Factor (SF)', color='#94a3b8', fontsize=9)
ax1.set_ylabel('Taxa de Transmissão (kbps)', color='#8b5cf6', fontsize=9)
ax2.set_ylabel('Sensibilidade do Receptor (dBm)', color='#22d3ee', fontsize=9)
ax1.tick_params(colors='#94a3b8'); ax2.tick_params(colors='#94a3b8')
ax1.grid(True, color='#3a3275', alpha=0.5, ls=':')
lines = p1 + p2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, facecolor='#241f4d', edgecolor='#3a3275', labelcolor='#ffffff', fontsize=8)
plt.title('Compromisso entre Taxa de Dados e Sensibilidade em LoRaWAN', color='#ffffff', fontsize=10.5, fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
