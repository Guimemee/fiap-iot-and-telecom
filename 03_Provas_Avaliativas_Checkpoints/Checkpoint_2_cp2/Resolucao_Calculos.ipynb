**Título:** Protocolos Leves de Transporte de Mensagens: MQTT vs. CoAP

> **Enunciado Original:**
> Implementar e comparar o desempenho de rede e overhead de cabeçalho entre os protocolos de aplicação IoT: MQTT (baseado em TCP e arquitetura Publish/Subscribe) e CoAP (Constrained Application Protocol, baseado em UDP e RESTful Request/Response). Analisar os níveis de garantia de entrega de serviço QoS 0, 1 e 2 no MQTT e mensagens Confirmable (CON) no CoAP.

### Formulação Matemática Formal
$$\text{Eficiência} = \frac{\text{Payload Útil}}{\text{Payload Útil} + \text{Overhead Cabeçalhos}} \times 100\%$$


### Passo 1: Arquitetura de Camada de Transporte
MQTT utiliza TCP garantindo confiabilidade por fluxo ordenado, exigindo handshake de 3 vias (SYN, SYN-ACK, ACK) e keep-alive periódico. CoAP utiliza UDP com cabeçalho ultra-compacto de apenas 4 bytes fixos.


### Passo 2: Análise de Overhead de Bytes por Mensagem
Para um dado útil de 20 bytes: MQTT requer cabeçalho fixo (2B) + cabeçalho de tópico (ex: 14B) + TCP (20B) + IP (20B) = 56 bytes totais. CoAP requer cabeçalho CoAP (4B) + UDP (8B) + IP (20B) = 32 bytes totais.


### Passo 3: Comparação de Eficiência de Banda
Eficiência MQTT = \frac{20}{56} \approx 35,7\%. Eficiência CoAP = \frac{20}{32} \approx 62,5\%. O CoAP é 75% mais eficiente no uso de bytes transmitidos pelo rádio.


### Passo 4: Qualidade de Serviço (QoS) e Resiliência
No MQTT: QoS 0 (no máximo uma vez), QoS 1 (pelo menos uma vez com PUBACK) e QoS 2 (exatamente uma vez com handshake de 4 mensagens). No CoAP: Mensagens Confirmable exigem ACK; Non-confirmable operam sem retorno.


### Passo 5: Diretriz de Escolha de Engenharia
MQTT é ideal para comunicação com brokers em nuvem e dispositivos permanentemente conectados à internet; CoAP é ideal para dispositivos ultra-restritos a bateria com sono profundo constante.


```python
import numpy as np
import matplotlib.pyplot as plt

cargas = np.array([5, 10, 20, 50, 100, 200]) # Payload util em bytes
overhead_mqtt = 36 # Overhead de cabeçalho MQTT + TCP
overhead_coap = 12 # Overhead CoAP + UDP
ef_mqtt = (cargas / (cargas + overhead_mqtt)) * 100
ef_coap = (cargas / (cargas + overhead_coap)) * 100
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#181436')
ax.set_facecolor('#100d22')
ax.plot(cargas, ef_coap, marker='o', color='#22d3ee', lw=2.5, label='CoAP / UDP (Alta Eficiência)')
ax.plot(cargas, ef_mqtt, marker='s', color='#8b5cf6', lw=2.5, label='MQTT / TCP (Orientado a Conexão)')
ax.set_title('Eficiência de Banda vs. Tamanho da Carga Útil (Payload)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tamanho da Carga Útil dos Sensores (Bytes)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Eficiência de Transmissão (%)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#3a3275', alpha=0.5, ls=':')
ax.legend(facecolor='#241f4d', edgecolor='#3a3275', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
