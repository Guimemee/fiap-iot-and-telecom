> Dimensionar os parâmetros de uma rede privada 5G Standalone (5G SA) com computação de borda (Multi-Access Edge Computing - MEC) operando em onda milimétrica (FR2 a 28 GHz) para controle de veículos autônomos industriais (AGVs). Configurar o fatiamento de rede (Network Slicing) garantindo isolamento da fatia URLLC (Ultra-Reliable Low-Latency Communication) com confiabilidade 99,999% e latência máxima de 1 ms.

### Formulação Matemática Formal
$$\text{Disponibilidade URLLC} \ge 1 - 10^{-5} = 99,999\%, \quad \text{Latência Interface Aérea} \le 0,5 \text{ ms}$$


### Passo 1: Pilares do 5G e Seleção da Fatiamento
O 5G divide o espectro em 3 pilares: eMBB (Banda Larga Móvel Aumentada), mMTC (Comunicações Massivas Tipo Máquina) e URLLC (Comunicações Ultra-Confiáveis de Baixa Latência). AGVs industriais demandam URLLC.


### Passo 2: Redução do Tempo de Intervalo de Transmissão (TTI)
No LTE 4G, o subquadro padrão tem duração fixa de 1 ms. No 5G NR com espaçamento de subportadoras (SCS) expandido para 60 kHz ou 120 kHz, o slot dura apenas 0,25 ms ou 0,125 ms, permitindo mini-slots de altíssima velocidade.


### Passo 3: Arquitetura MEC no Ponto de Acesso Local
Instalação do plano de controle do usuário (UPF - User Plane Function) e servidores de inferência na própria subestação fabril, eliminando a viagem de dados (Backhaul) até a nuvem pública e garantindo latência RTT < 3 ms.


### Passo 4: Garantia de Confiabilidade com Diversidade Espacial
MIMO Massivo e transmissão redundante de pacotes críticos em portadoras duplas garantem que a probabilidade de falha por pacote permaneça abaixo de 10^{-5}.


### Passo 5: Conclusão de Eficiência Operacional
Eliminação completa de cabeamentos físicos e cabos umbilicais, permitindo reconfiguração dinâmica em minutos de rotas de frotas de 60 robôs industriais móveis.


```python
import numpy as np
import matplotlib.pyplot as plt

dist_agv = np.linspace(5, 120, 100) # Distancia da antena 5G em metros
# Latencia RTT comparativa
lat_wifi = 15.0 + 0.12 * dist_agv + np.random.RandomState(42).normal(0, 1.5, 100)
lat_5g_urllc = 1.2 + 0.005 * dist_agv
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#181436')
ax.set_facecolor('#100d22')
ax.plot(dist_agv, lat_wifi, color='#ef4444', lw=1.8, label='Wi-Fi Industrial Legado (Jitter Alto)')
ax.plot(dist_agv, lat_5g_urllc, color='#10b981', lw=2.5, label='5G URLLC Standalone + Edge MEC')
ax.axhline(5.0, color='#f59e0b', ls='--', lw=1.5, label='Limite Crítico de Segurança AGV (5 ms)')
ax.set_title('Latência de Controle RTT: 5G URLLC vs. Wi-Fi Industrial', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Distância do Robô até o Ponto de Acesso (Metros)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Latência de Ida e Volta RTT (Milissegundos)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#3a3275', alpha=0.5, ls=':')
ax.legend(facecolor='#241f4d', edgecolor='#3a3275', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
